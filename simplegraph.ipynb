```python
from random import shuffle
import csv
```

## Set the names and the output size of the graph

```python
m_names = ["Oliver", "Noah", "Liam", "Elijah", "Mason", "James", "Benjamin", "Lucas", "Henry", "Alexander",             
           "Jackson", "Sebastian", "Daniel", "Matthew", "Samuel", "David", "Joseph", "John", "Levi", "Owen",             
           "Michael", "William", "Ethan", "Joshua", "Andrew", "Nathan", "Christopher", "Gabriel", "Isaac",             
           "Ryan", "Jonathan", "Logan", "Caleb", "Christian", "Thomas", "Adrian", "Hunter", "Charles", "Connor"] 
f_names = [ "Amelia", "Isabella", "Ava", "Sophia", "Mia", "Charlotte", "Harper", "Evelyn", "Abigail", "Ella", 
            "Grace", "Aria", "Chloe", "Luna", "Penelope", "Mila", "Aurora", "Hazel", "Elizabeth", 
           "Emily", "Victoria", "Madison", "Lily", "Sofia", "Riley", "Layla", "Audrey", "Leah", "Nora", 
           "Zoey", "Hannah", "Addison", "Stella", "Natalie", "Zoe", "Paisley", "Ellie", "Savannah", "Anna" ]
```

```python
# Set the number of families in each generation
fam_num = 2 # Min 2, max 12
```

## Generate the relations and questions

```python
# Shuffle the lists of names to create a different graph each time
shuffle(m_names)
shuffle(f_names)
```

```python
# Separate names into generations
gen_0_m = m_names[0:fam_num]
gen_0_f = f_names[0:fam_num]
gen_1_m = m_names[fam_num + 1:2*fam_num + 1]
gen_1_f = f_names[fam_num + 1:2*fam_num + 1]
gen_2_m = m_names[2*fam_num + 2:3*fam_num + 2]
gen_2_f = f_names[2*fam_num + 2:3*fam_num + 2]

# Create pairs of siblings
gen_0_sibs = list(zip(gen_0_f, gen_0_m))
gen_1_sibs = list(zip(gen_1_f, gen_1_m))
gen_2_sibs = list(zip(gen_2_f, gen_2_m))

shuffle(gen_0_m)
shuffle(gen_0_f)
shuffle(gen_1_m)
shuffle(gen_1_f)
shuffle(gen_2_m)
shuffle(gen_2_f)

# Create pairs of spouses
gen_0_spouses = list(zip(gen_0_f, gen_0_m))
gen_1_spouses = list(zip(gen_1_f, gen_1_m))
gen_2_spouses = list(zip(gen_2_f, gen_2_m))

good = False
while good == False:
    good = True
    for spouses in gen_0_spouses:
        if spouses in gen_0_sibs:
            good = False
    if good == False:
        print('Try again')
        shuffle(gen_0_f)
        gen_0_spouses = list(zip(gen_0_f, gen_0_m))
    else:
        continue

good = False
while good == False:
    good = True
    for spouses in gen_1_spouses:
        if spouses in gen_1_sibs:
            good = False
    if good == False:
        print('Try again')
        shuffle(gen_1_f)
        gen_1_spouses = list(zip(gen_1_f, gen_1_m))
    else:
        continue

good = False
while good == False:
    good = True
    for spouses in gen_2_spouses:
        if spouses in gen_2_sibs:
            good = False
    if good == False:
        print('Try again')
        shuffle(gen_2_f)
        gen_2_spouses = list(zip(gen_2_f, gen_2_m))
    else:
        continue
```

```python
# Utility sets and functions
spouse_gens = [gen_0_spouses, gen_1_spouses, gen_2_spouses]
sib_gens = [gen_0_sibs, gen_1_sibs, gen_2_sibs]
spouse_gens_2 = [gen_0_spouses, gen_1_spouses]
sib_gens_2 = [gen_1_sibs, gen_2_sibs]

id_pairs = list(enumerate(gen_1_spouses))

def get_index(name):
    for index, pair in id_pairs:
        if name in pair:
            return index
    return -1
```

```python
# Create the relations and the questions
relations = []
questions = []

for gen in spouse_gens:
    for i in range(fam_num):
        relations.append(f'{gen[i][0]} is the spouse of {gen[i][1]}.')
        relations.append(f'{gen[i][1]} is the spouse of {gen[i][0]}.')
        questions.append((f'Who is the spouse of {gen[i][1]}?', gen[i][0]))
        questions.append((f'Who is the spouse of {gen[i][0]}?', gen[i][1]))

for gen in sib_gens:
    for i in range(fam_num):
        relations.append(f'{gen[i][0]} is the sibling of {gen[i][1]}.')
        relations.append(f'{gen[i][1]} is the sibling of {gen[i][0]}.')
        questions.append((f'Who is the sibling of {gen[i][1]}?', gen[i][0]))
        questions.append((f'Who is the sibling of {gen[i][0]}?', gen[i][1]))

for i in range(2):
    parents = spouse_gens_2[i]
    children = sib_gens_2[i]
    for j in range(fam_num):
        for k in range(2): # Gender of parent, 0 is female, 1 is male
            relations.append(f'{children[j][0]} is the daughter of {parents[j][k]}.')
            relations.append(f'{children[j][1]} is the son of {parents[j][k]}.')
            if k == 0:
                relations.append(f'{parents[j][k]} is the mother of {children[j][0]}.')
                relations.append(f'{parents[j][k]} is the mother of {children[j][1]}.')
                questions.append((f'Who is the mother of {children[j][0]}?', parents[j][k]))
                questions.append((f'Who is the mother of {children[j][1]}?', parents[j][k]))
            else:
                relations.append(f'{parents[j][k]} is the father of {children[j][0]}.')
                relations.append(f'{parents[j][k]} is the father of {children[j][1]}.')
                questions.append((f'Who is the father of {children[j][0]}?', parents[j][k]))
                questions.append((f'Who is the father of {children[j][1]}?', parents[j][k]))
            questions.append((f'Who is the daughter of {parents[j][k]}?', children[j][0]))
            questions.append((f'Who is the son of {parents[j][k]}?', children[j][1]))

for i in range(fam_num): # i traverses the list of generation 1 siblings
    idx = get_index(gen_1_sibs[i][0]) # For the wife in gen 1 spouses pair i, what is the gen 1 spouses index?
    for j in range(2):
        if j == 0:
            relations.append(f'{gen_0_spouses[i][j]} is the grandmother of {gen_2_sibs[idx][0]}.')
            relations.append(f'{gen_0_spouses[i][j]} is the grandmother of {gen_2_sibs[idx][1]}.')
            questions.append((f'Who is the grandmother of {gen_2_sibs[idx][0]}?', gen_0_spouses[i][j]))
            questions.append((f'Who is the grandmother of {gen_2_sibs[idx][1]}?', gen_0_spouses[i][j]))
        else:
            relations.append(f'{gen_0_spouses[i][j]} is the grandfather of {gen_2_sibs[idx][0]}.')
            relations.append(f'{gen_0_spouses[i][j]} is the grandfather of {gen_2_sibs[idx][1]}.')
            questions.append((f'Who is the grandfather of {gen_2_sibs[idx][0]}?', gen_0_spouses[i][j]))
            questions.append((f'Who is the grandfather of {gen_2_sibs[idx][1]}?', gen_0_spouses[i][j]))
        relations.append(f'{gen_1_sibs[i][1]} is the uncle of {gen_2_sibs[idx][j]}.')
        questions.append((f'Who is the uncle of {gen_2_sibs[idx][j]}?', gen_1_sibs[i][1]))
    idx = get_index(gen_1_sibs[i][1])
    for j in range(2):
        if j == 0:
            relations.append(f'{gen_0_spouses[i][j]} is the grandmother of {gen_2_sibs[idx][0]}.')
            relations.append(f'{gen_0_spouses[i][j]} is the grandmother of {gen_2_sibs[idx][1]}.')
            questions.append((f'Who is the grandmother of {gen_2_sibs[idx][0]}?', gen_0_spouses[i][j]))
            questions.append((f'Who is the grandmother of {gen_2_sibs[idx][1]}?', gen_0_spouses[i][j]))
        else:
            relations.append(f'{gen_0_spouses[i][j]} is the grandfather of {gen_2_sibs[idx][0]}.')
            relations.append(f'{gen_0_spouses[i][j]} is the grandfather of {gen_2_sibs[idx][1]}.')
            questions.append((f'Who is the grandfather of {gen_2_sibs[idx][0]}?', gen_0_spouses[i][j]))
            questions.append((f'Who is the grandfather of {gen_2_sibs[idx][1]}?', gen_0_spouses[i][j]))
        relations.append(f'{gen_1_sibs[i][0]} is the aunt of {gen_2_sibs[idx][j]}.')
        questions.append((f'Who is the aunt of {gen_2_sibs[idx][j]}?', gen_1_sibs[i][0]))
```

```python
# Make questions unique
newqs = sorted(questions)
topop = []
for i in range(len(newqs)-1):
    if newqs[i][0] == newqs[i+1][0]:
        newqs[i] = (newqs[i][0], newqs[i][1], newqs[i+1][1])
        topop.append(i+1)
for num in topop:
    newqs.pop(num)

questions = newqs
```

## Write the graph and questions to CSV files

```python
# Write the relations and questions to a CSV

## Total graph
graph_num = 1
rel_out = f'./simplegraphs/{graph_num}-rels.csv'
quest_out = f'./simplegraphs/{graph_num}-questions.csv'

with open(rel_out, 'w', newline='') as file:
    writer = csv.writer(file)
    for rel in sorted(relations):
        writer.writerow([rel])

with open(quest_out, 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(sorted(questions))

## Grandparent Graph
```
